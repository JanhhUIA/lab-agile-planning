```python
import pandas as pd
import matplotlib.pyplot as plt

# Load the CSV file
df = pd.read_csv('IEA_Reference_15MW_240.csv')  # No longer skipping the header row

# Extract columns
wind_speeds = df['Wind Speed [m/s]'].values
power_kw = df['Power [kW]'].values
ct = df['Ct [-]'].values

# Convert power to MW for nicer plot
power_mw = power_kw / 1000

# Plot power and Ct curves
plt.figure(figsize=(10, 6))
plt.plot(wind_speeds, power_mw, 'b-', linewidth=2, label='Power (MW)')
plt.xlabel('Wind Speed (m/s)')
plt.ylabel('Power (MW)', color='blue')
plt.tick_params(axis='y', labelcolor='blue')
plt.grid(True, alpha=0.3)
plt.xlim(0, 25) # Set x-axis limit from 0 to 25 m/s

# Second y-axis for Ct
ax2 = plt.twinx()
ax2.plot(wind_speeds, ct, 'r--', linewidth=2, label='Ct')
ax2.set_ylabel('Thrust Coefficient Ct (-)', color='red')
ax2.tick_params(axis='y', labelcolor='red')

plt.title('IEA 15 MW Reference Turbine - Power and Thrust Coefficient Curves')
plt.legend(loc='upper left')
plt.tight_layout()
plt.show()

print(f"Rated power: {power_mw.max():.1f} MW")
print(f"Number of data points: {len(wind_speeds)}")
```

```python
from py_wake.wind_turbines import WindTurbines
from py_wake.wind_turbines.power_ct_functions import PowerCtTabular

# Define the power and Ct function from the loaded data
power_ct_function = PowerCtTabular(
    ws=wind_speeds,          # wind speeds in m/s
    power=power_kw,          # power in kW
    power_unit='kW',
    ct=ct                    # thrust coefficient
)

# Project specifications
num_turbines = 15                  # 15 × 15 MW = 225 MW farm
rotor_diameter = 240               # meters
hub_height = 150                   # meters → tip height = 150 + 120 = 270 m < 280 m limit ✓

# Create the WindTurbines object
wt = WindTurbines(
    names=['IEA_15MW'],
    diameters=[rotor_diameter],
    hub_heights=[hub_height],
    powerCtFunctions=[power_ct_function]
)

# Display basic info
print(wt)
print(f"\nTurbine: IEA 15 MW")
print(f"Rotor diameter: {rotor_diameter} m")
print(f"Hub height: {hub_height} m (max tip height 270 m)")
print(f"Number of turbines in farm: {num_turbines}")
```

```python
from py_wake.site import UniformWeibullSite
from scipy.special import gamma  # Correct import for gamma function
import numpy as np

# Realistic Weibull parameters for Utsira Nord
wd = [0, 90, 180, 270]                    # Directions: N, E, S, W
f = [0.35, 0.20, 0.20, 0.25]               # Sector probabilities
A = [9.18, 9.78, 9.53, 9.91]               # Weibull scale A
k = [2.39, 2.45, 2.41, 2.59]               # Weibull shape k

site = UniformWeibullSite(
    p_wd=f,
    a=A,
    k=k,
    ti=0.08
)

# Manual mean wind speed calculation (Weibull mean = A * Γ(1 + 1/k))
mean_ws_sector = np.array(A) * gamma(1 + 1/np.array(k))
overall_mean_ws = np.sum(np.array(f) * mean_ws_sector)

print(f"Site: Utsira Nord (4-sector UniformWeibull model)")
print(f"Overall mean wind speed at hub height: {overall_mean_ws:.2f} m/s")
```

```python
import utm
from shapely.geometry import Polygon
import matplotlib.pyplot as plt
import numpy as np

# Smaller compliant GPS coordinates (quadrilateral, max edge <9 km)
gps_coords = [
    (59.28, 4.75),   # Southwest
    (59.32, 4.75),   # Northwest
    (59.32, 4.85),   # Northeast
    (59.28, 4.85)    # Southeast
]

# Convert to UTM
utm_coords = [utm.from_latlon(lat, lon)[:2] for lat, lon in gps_coords]

print("Boundary corners (UTM easting, northing):")
for i, (e, n) in enumerate(utm_coords):
    print(f"  Point {i+1}: ({e:.0f}, {n:.0f}) m")

# Area calculation
poly = Polygon(utm_coords)
area_km2 = poly.area / 1e6

# Max edge
distances = []
for i in range(len(utm_coords)):
    for j in range(i+1, len(utm_coords)):
        dx = utm_coords[i][0] - utm_coords[j][0]
        dy = utm_coords[i][1] - utm_coords[j][1]
        dist_km = np.sqrt(dx**2 + dy**2) / 1000
        distances.append(dist_km)
max_edge_km = max(distances)

print(f"\nWind farm area: {area_km2:.2f} km²")
print(f"Maximum edge length: {max_edge_km:.2f} km (≤ 9 km ✓)")

# Fixed plot (no TypeError)
utm_coords_closed = utm_coords + [utm_coords[0]]  # Properly close
bx, by = zip(*utm_coords_closed)

plt.figure(figsize=(8, 8))
plt.plot(bx, by, 'k-', linewidth=3, label='Farm boundary')
plt.scatter([p[0] for p in utm_coords], [p[1] for p in utm_coords], c='red', s=100, zorder=5)
plt.xlabel('Easting (m)')
plt.ylabel('Northing (m)')
plt.title('Compliant Wind Farm Boundary in Utsira Nord')
plt.grid(True, alpha=0.3)
plt.axis('equal')
plt.legend()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Grid: 5 rows × 3 columns = 15 turbines
D = rotor_diameter  # 240 m
spacing_downwind = 8 * D  # 1920 m
spacing_crosswind = 5 * D  # 1200 m

# Grid positions
n_rows = 5
n_cols = 3
x_pos = np.linspace(-spacing_downwind, spacing_downwind, n_cols)
y_pos = np.linspace(-2*spacing_crosswind, 2*spacing_crosswind, n_rows)

x_grid, y_grid = np.meshgrid(x_pos, y_pos)
x_initial = x_grid.flatten()
y_initial = y_grid.flatten()

# Center the grid
center_x = np.mean([c[0] for c in utm_coords])
center_y = np.mean([c[1] for c in utm_coords])
x_initial += center_x
y_initial += center_y

# Fixed plotting (close polygon properly)
utm_coords_closed = utm_coords + [utm_coords[0]]
bx, by = zip(*utm_coords_closed)

plt.figure(figsize=(10,10))
plt.plot(bx, by, 'k-', linewidth=3, label='Boundary')
plt.scatter(x_initial, y_initial, c='cyan', s=150, edgecolor='black', label='Initial turbines (5×3 grid)')
plt.xlabel('Easting (m)')
plt.ylabel('Northing (m)')
plt.title('Initial Grid Layout (15 × IEA 15 MW Turbines)')
plt.axis('equal')
plt.grid(True, alpha=0.3)
plt.legend()
plt.show()

print("Initial layout created: 15 turbines centered in boundary")
```

```python
from py_wake.literature.gaussian_models import Bastankhah_PorteAgel_2014  # Correct import for Bastankhah Gaussian

# Use the literature model with k=0.10
wfm = Bastankhah_PorteAgel_2014(
    site=site,
    windTurbines=wt,
    k=0.10
)

# Simulate using the turbine's wind speeds from CSV to ensure overlap
sim_res = wfm(x=x_initial, y=y_initial, ws=wind_speeds)

# AEP in GWh (now correctly integrated)
aep_gwh_initial = sim_res.aep().sum(['wt','wd','ws']).values.item()

# Metrics
gross_aep_gwh = num_turbines * 15 * 8.76
capacity_factor = aep_gwh_initial / gross_aep_gwh
wake_loss_pct = 100 * (1 - capacity_factor)
aep_per_area_initial = aep_gwh_initial / area_km2

print("=== INITIAL RESULTS ===")
print(f"Annual Energy Production: {aep_gwh_initial:.1f} GWh/year")
print(f"AEP per km²: {aep_per_area_initial:.1f} GWh/km²/year")
print(f"Capacity factor: {capacity_factor:.1%}")
print(f"Wake losses: {wake_loss_pct:.1f}%")

# Wake map
sim_res.flow_map(wd=270, ws=10.0).plot_wake_map()
plt.title("Initial Layout Wake Map (West Wind, 10 m/s)")
plt.show()
```

```python
# Use a wind speed that exactly matches one of the values in sim_res.ws
# Find the wind speed closest to 10 m/s in the simulation results
target_ws = 10.0
sim_ws_values = sim_res.ws.values
# Find the index of the wind speed closest to 10.0
idx = np.argmin(np.abs(sim_ws_values - target_ws))
selected_ws = sim_ws_values[idx]

print(f"Plotting wake map for wind speed: {selected_ws} m/s")
sim_res.flow_map(wd=270, ws=selected_ws).plot_wake_map()
plt.title(f"Initial Layout Wake Map (West Wind, {selected_ws:.1f} m/s)")
plt.show()
```

```python
from topfarm import TopFarmProblem
from topfarm.cost_models.cost_model_wrappers import CostModelComponent
from topfarm.constraint_components.boundary import XYBoundaryConstraint
from topfarm.easy_drivers import EasyScipyOptimizeDriver
from topfarm.plotting import XYPlotComp

boundary_constraint = XYBoundaryConstraint(utm_coords)

def calc_aep(x, y):
    sim = wfm(x=x, y=y, ws=wind_speeds)
    return -sim.aep().sum(['wt','wd','ws']).values.item()

cost_comp = CostModelComponent(
    input_keys=[('x', x_initial), ('y', y_initial)],
    n_wt=num_turbines,
    cost_function=calc_aep,
    output_keys="aep_cost",
    objective=True
)

problem = TopFarmProblem(
    design_vars={'x': x_initial, 'y': y_initial},
    cost_comp=cost_comp,
    constraints=[boundary_constraint],
    driver=EasyScipyOptimizeDriver(optimizer='SLSQP', maxiter=200),
    plot_comp=XYPlotComp()
)

cost, state, recorder = problem.optimize()

aep_gwh_opt = -cost
improvement_pct = 100 * (aep_gwh_opt / aep_gwh_initial - 1)
final_aep_per_area = aep_gwh_opt / area_km2

print("\n=== OPTIMIZED RESULTS ===")
print(f"Optimized AEP: {aep_gwh_opt:.1f} GWh/year")
print(f"Improvement: +{improvement_pct:.1f}%")
print(f"Final AEP per km²: {final_aep_per_area:.1f} GWh/km²/year")

utm_coords_closed = utm_coords + [utm_coords[0]]
bx, by = zip(*utm_coords_closed)
plt.figure(figsize=(10,10))
plt.plot(bx, by, 'k-', linewidth=3)
plt.scatter(state['x'], state['y'], c='lime', s=150, edgecolor='black')
plt.title('Optimized Wind Farm Layout')
plt.axis('equal')
plt.legend(['Boundary', 'Optimized turbines'])
plt.grid(True)
plt.show()
```
